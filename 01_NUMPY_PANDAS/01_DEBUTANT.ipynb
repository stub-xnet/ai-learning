## Correction — Exercice 9 : Vue ou copie ? Prédire

````python
a = np.arange(6); b = a[1:4]; b[0] = 99; print(1, a)
a = np.arange(6); b = a[[1, 2, 3]]; b[0] = 99; print(2, a)
a = np.arange(6); b = a[a > 2]; b[0] = 99; print(3, a)
a = np.arange(6); b = a.reshape(2, 3); b[0, 0] = 99; print(4, a)
a = np.arange(6); b = a.copy(); b[0] = 99; print(5, a)
````
